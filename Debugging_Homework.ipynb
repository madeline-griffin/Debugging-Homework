```python
## 1a
def wrong_add_function(arg1,arg2):
    arg1_index=0
    while arg1_index < len(arg1):
        arg_2_sum = 0
        for arg2_elements in arg2:
            arg_2_sum = sum([arg1[arg1_index]+i for i in arg2])
        # 1.a: the error is in this loop. arg1[arg1_index] gets added once for
        # every element of arg2 (3 times here) instead of once.
        print(f"We are making an error in the loop: arg_2_sum is {arg_2_sum} "
               f"for arg1[{arg1_index}]. The correct answer is supposed to be: "
               f"{arg1[arg1_index]+arg2[arg1_index]}")
        arg1[arg1_index]=arg_2_sum  
        arg1_index+=1
    return arg1

arg1 = [1,2,3]
arg2 = [1,1,1]

Answer_1a= wrong_add_function(arg1, arg2)
print(Answer_1a)
```

```python
## 1b
def correct_add_function(arg1,arg2):
    arg1_index=0
    while arg1_index < len(arg1):
      arg_2_sum = arg1[arg1_index]+arg2[arg1_index]   # the fix
      arg1[arg1_index]=arg_2_sum  
      arg1_index+=1
    return arg1
arg1 = [1,2,3]
arg2 = [1,1,1]

Answer_1b= correct_add_function(arg1, arg2)
print(Answer_1b)
```

```python
## 2a
def wrong_add_function(arg1,arg2):
    #numeric section
    if sum([type(i)==int for i in arg1])==len(arg1) and \
    sum([type(i)==int for i in arg2])==len(arg2):
        arg1_index=0
        while arg1_index < len(arg1):
            arg_2_sum = arg1[arg1_index]+arg2[arg1_index]   # 2.a: fix from 1.b
            arg1[arg1_index]=arg_2_sum  
            arg1_index+=1
        return arg1
    #string section
    elif sum([type(i)==str for i in arg1])==len(arg1) and \
    sum([type(i)==str for i in arg2])==len(arg2):
        arg1_index=0
        while arg1_index < len(arg1):
            arg_2_sum = ''
            for arg2_elements in arg2:
                arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+str(arg_2_sum)
            arg1_index+=1
        return arg1
```

```python
## 2b
def exception_add_function(arg1,arg2):
    try:
        result = wrong_add_function(arg1,arg2)
        if result is None:
            raise TypeError("mixed or unsupported element types")
        return result
    except TypeError:
        expected_type = str if (len(arg1) > 0 and type(arg1[0]) == str) else int
        for arg_number, arg in enumerate([arg1, arg2], start=1):
            for n, element in enumerate(arg):
                if type(element) != expected_type:
                    return (f"Your input argument {arg_number} at element {n} is not of the expected type. Please change this and rerun.")
        return "Your input is not of the expected type. Please change this and rerun."
arg_str_1=['1','2','3']
arg_str_2=['1','1', 1]
Answer_2b= exception_add_function(arg_str_1,arg_str_2)
print(Answer_2b)
```

```python
## 2c
def correction_add_function(arg1,arg2):
    try:
        result = wrong_add_function(list(arg1), list(arg2))
        if result is None:
            raise TypeError("mixed or unsupported element types")
        return result
    except TypeError:
        return wrong_add_function([str(i) for i in arg1], [str(i) for i in arg2])
arg_str_1=['1','2','3']
arg_str_2=['1','1', 1]
Answer_2c= correction_add_function(arg_str_1,arg_str_2)
print(Answer_2c)
```
